```python
from torchvision.datasets import FashionMNIST
```

```python
fm_train = FashionMNIST(root='./dataSet', train=True, download=True)
fm_test = FashionMNIST(root='./dataSet', train=False, download=True)
```

```python
train_input = fm_train.data
train_target = fm_train.targets
```

```python
train_scaled = train_input / 255.0
```

```python
from sklearn.model_selection import train_test_split
```

```python
train_scaled, val_scaled, train_target, val_target = train_test_split(
    train_scaled, train_target, test_size=0.2, random_state=42
)
```

```python
f'{train_scaled.shape} {val_scaled.shape}'
```

```python
import torch.nn as nn
```

```python
model = nn.Sequential(
    nn.Flatten(),
    nn.Linear(784, 100),
    nn.ReLU(),
    nn.Linear(100, 10)
)
```

```python
from torchinfo import summary
```

```python
summary(model, input_size=(32, 28,28))
```

```python
import torch
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)
```

```python
import torch.optim as optim
```

```python
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters())
```

```python
epochs = 5
batches = int(len(train_scaled)/32)
for epoch in range(epochs):
    model.train()
    train_loss = 0
    for i in range(batches):
        inputs = train_scaled[i*32:(i+1)*32].to(device)
        targets =train_target[i*32:(i+1)*32].to(device)

        optimizer.zero_grad()

        outputs = model(inputs)
        loss = criterion(outputs, targets)
        loss.backward()
        optimizer.step()
        train_loss += loss.item()
    print(f"에포크: {epoch+1}, 손실: {train_loss/batches:.4f}")
```

```python
model.eval()
# 모델을 '평가 모드(evaluation mode)'로 전환합니다.
# Dropout이나 BatchNorm처럼 학습할 때와 평가할 때 다르게 동작하는 레이어들에게
# "지금은 검증 중이니 평가용 동작을 수행해라"라고 알립니다.

with torch.no_grad():
# 이 블록 내부에서 일어나는 모든 연산에 대해 기울기(Gradient) 추적을 비활성화합니다.
# 검증 단계에서는 역전파(Backpropagation)로 가중치를 업데이트할 필요가 없으므로,
# 메모리 사용량을 줄이고 연산 속도를 대폭 향상시킵니다.

    val_scaled = val_scaled.to(device)
    # 정규화된 검증 입력 데이터(X)를 모델이 위치한 장치(CPU 또는 GPU/CUDA)로 이동시킵니다.
    # 모델과 입력 데이터가 같은 장치에 있어야 연산이 가능합니다.

    val_target = val_target.to(device)
    # 검증 데이터의 정답 레이블(y)을 동일한 장치(CPU 또는 GPU)로 이동시킵니다.

    outputs = model(val_scaled)
    # 모델에 입력 데이터를 넣어 예측 결과(Logits 또는 확률값)를 출력합니다.
    # 결과 형태는 [배치 크기, 클래스 개수]의 2차원 텐서가 됩니다.

    predicts = torch.argmax(outputs, 1)
    # outputs의 1번 차원(클래스 축)을 기준으로 가장 높은 값을 가진 인덱스를 선택합니다.
    # 즉, 모델이 생각하는 가장 확률이 높은 클래스 번호(예측값)를 추출합니다.

    corrects = (predicts == val_target).sum().item()
    # (predicts == val_target): 예측값과 정답을 비교해 같으면 True(1), 다르면 False(0)로 구성된 텐서를 만듭니다.
    # .sum(): True의 개수(맞춘 개수)를 모두 더합니다.
    # .item(): 텐서 형태로 존재하는 결과에서 순수 파이썬 숫자(int) 값만 뽑아냅니다.

    print(f'outputs: {outputs[[0]]}\npredicts: {predicts[0]}\nval_target: {val_target[0]}\ncorrects: {corrects}')

accuarcy = corrects / len(val_target)
# 맞춘 개수(corrects)를 전체 검증 데이터 개수(len(val_target))로 나누어 정확도를 계산합니다.

print(f"검증 정확도: {accuarcy:.4f}")
# 계산된 정확도를 소수점 넷째 자리까지 포맷팅하여 화면에 출력합니다.
```
